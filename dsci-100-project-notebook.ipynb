The stacked histograms below depict the distribution of regional tournament win percentage for each tournament finish category. As we can see, the tournament win percentages of teams that were eliminate in the first, second, or the top 16 rounds was lower on average than those that finished in higher positions in the tournament. Thus, if a new observation/team ended up having a lower regional tournament win percentage, it would be less likely that that team would place "well" (move past the top 16 round) in the tournament. 

```r
options(repr.plot.width = 12, repr.plot.height = 12)
march_madness_reg_wins_plot_by_tourney_finish <- march_madness_tidy_data |> ggplot(aes(x = reg_wins_pct, fill = as_factor(tourney_finish))) + geom_histogram(binwidth = 3) + 
                                                labs(x = "Regional Tournament Win Percentage", y = "Frequency", fill = "Tournament Result") + 
                                                ggtitle("Frequency of Teams' Regional Tournament Win Percentage")+ 
                                                theme(plot.title = element_text(size = 24, hjust = 0.5), axis.title = element_text(size = 18, hjust = 0.5)) +
                                                facet_grid(rows = vars(as_factor(tourney_finish)))
                                
march_madness_reg_wins_plot_by_tourney_finish
```

I plan to build a K-nearest-neighbors model for classification to predict the tournament finish of a team based on its seed placement and regional win percentage. We use KNN because we are trying to use two quantitative variables, the seed placement and regional win percentage of a team, to predict a categorical variable, the tournament finish category of the team.  

First, to find which value of k is most ideal for our KNN model, we would randomly split our dataset into training data (75%) and testing data (25%). Since there is significant class imbalance, with many more observations being teams that lost on the first round than those that made it farther in the tournament, we would need to perform oversampling on rarer classes in our training data in order to balance each category. 

Then, we will further split that training data into 5 folds for cross-validation, tuning our model from k = 1-50. For each value of k, we will treat one fold as a validation set (rest as training) and assessing how well the model can correctly predict the tournament result. We will repeat this process for all 5 folds of our training data, and proceed to average the accuracies of this process on our validation sets. 

The value of k that yields the highest average accuracy in this cross-validation process would be used on our testing set, where we would investigate the model's accuracy and precision. High accuracy is important because we want to maximize the total number of correct predictions for a team with a specific seed and regional percentage.

One limitation of our KNN model is its high computation power. Our march madness dataset is quite large (over 2000 observations) and using our model to create predictions on our test set may be a slower process.
